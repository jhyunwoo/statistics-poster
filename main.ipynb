```python
f=open("법정동코드 전체자료.txt", 'r', encoding="cp949")

area_code = []

while True:
    line=f.readline()
    if not line:break
    code = line[0:5]
    split = line.split()
    if split:
        area_name = split[1]+" "+split[2]
        if code not in area_code:
            area_code.append({code:code, area_name:area_name})

f.close

print(area_code)
```
