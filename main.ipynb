```python
f=open("법정동코드 전체자료.txt", 'r', encoding="cp949")

area_code = []

while True:
    line=f.readline()
    if not line:break
    code = line[0:5]
    split = line.split()
    if split:
        if len(split)>3:
            area_name = split[1]+" "+split[2]
            if {"code":code, "area_name":area_name} not in area_code:
                area_code.append({"code":code, "area_name":area_name})

f.close

print(area_code)
```

```python
month_list = []
for i in range(2016, 2023):
    for j in range(1, 13):
        if j< 10:
            j = "0"+str(j)
        month_list.append(str(i)+str(j))
```

```python
print(month_list)
```

```python
full_df = pd.DataFrame()

for i in area_code:
    for j in month_list:
        params ={'serviceKey' : 'oSDE9yxibBFt0A4pLMkWeJhdtLACtUJbxNKQIZf7CW2RZrR8OE2uyOxfgOYPyPQDvxQ7cM/EWsGwOWuCRgvtBQ==', 'LAWD_CD' : i["code"], 'DEAL_YMD' : j }
        response = requests.get(url, params=params)
        xml_data = response.text
        json_str = json.dumps(xmltodict.parse(xml_data), indent=4)
        dict = json.loads(json_str)
        json_data=dict["response"]["body"]["items"]["item"]
        json_data=json.dumps(json_data)
        df = pd.DataFrame(json.loads(json_data)) 
        full_df = pd.concat([full_df, df])
    full_df.to_csv(f'{i["area_name"]}.csv', header=True, index=True,encoding='ms949') 
    full_df = pd.DataFrame()
```
